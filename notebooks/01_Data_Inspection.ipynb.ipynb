```python
features = pd.read_csv("../data/raw/features.csv")
stores = pd.read_csv("../data/raw/stores.csv")
```

```python
print("FEATURES")
print(features.shape)
display(features.head())

print("\nSTORES")
print(stores.shape)
display(stores.head())
```

```python
stores.info()
```

```python
sales = train.copy()
```

```python
sales['Date'] = pd.to_datetime(sales['Date'])
features['Date'] = pd.to_datetime(features['Date'])
```

```python
print(sales['Date'].dtype)
print(features['Date'].dtype)
```

```python
features.columns.tolist()
```

```python
features_for_merge = features.drop(columns=['IsHoliday'])
```

```python
sales = sales.merge(
    features_for_merge,
    on=['Store', 'Date'],
    how='left'
)
```

```python
print("Rows:", sales.shape[0])
print("Columns:", sales.shape[1])
```

```python
sales.head()
```

```python
sales = sales.merge(
    stores,
    on='Store',
    how='left'
)
```

```python
print("Rows:", sales.shape[0])
print("Columns:", sales.shape[1])
```

```python
missing = sales.isnull().sum()

print(missing[missing > 0])
```

```python
markdown_cols = [
    'MarkDown1',
    'MarkDown2',
    'MarkDown3',
    'MarkDown4',
    'MarkDown5'
]

sales[markdown_cols] = sales[markdown_cols].fillna(0)
```

```python
sales[markdown_cols].isnull().sum()
```

```python
sales = sales.sort_values(['Store', 'Date'])

sales[['CPI', 'Unemployment']] = (
    sales.groupby('Store')[['CPI', 'Unemployment']]
         .transform(lambda x: x.ffill().bfill())
)
```

```python
sales[['CPI', 'Unemployment']].isnull().sum()
```

```python
print(sales.isnull().sum())
```

```python
print("Duplicate rows:", sales.duplicated().sum())
```

```python
sales['Year'] = sales['Date'].dt.year
sales['Month'] = sales['Date'].dt.month
sales['Month_Name'] = sales['Date'].dt.month_name()
sales['Quarter'] = sales['Date'].dt.quarter
sales['Week'] = sales['Date'].dt.isocalendar().week.astype(int)
```

```python
sales[['Date', 'Year', 'Month', 'Month_Name', 'Quarter', 'Week']].head()
```

```python
print("Final rows:", sales.shape[0])
print("Final columns:", sales.shape[1])
```
